```python
# Import dependencies
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
# Get the dataset
file_path = "petfinder-pawpularity-score/train.csv"
df = pd.read_csv(file_path)
data_np = df.iloc[:, 1:].to_numpy()

print(data_np.shape)
print(data_np[:5])
```

```python
# Histogram plot
plt.hist(data_np[:, -1], bins=20, color='skyblue', edgecolor='black', alpha=0.7)

plt.xlabel("Pawpularity Score")
plt.ylabel("Frequency")
plt.title("Distribution of Pawpularity Scores")

plt.show()
```
